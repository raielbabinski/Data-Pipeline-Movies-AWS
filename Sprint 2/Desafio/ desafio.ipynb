Importando as biblioteca Pandas e Matplotlib.

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

Criar o dataframe usando pandas.

```python
df = pd.read_csv('googleplaystore.csv')
df = df.drop((df.loc[df['App'] == 'Life Made WI-Fi Touchscreen Photo Frame'].index))
```

---

#### Desenvolvimento

---

##### **1)** Remova linhas duplicatas

```python
print(df.index)
df = df.drop_duplicates(subset='App', keep='first', inplace=False)
print(df.index)
```

##### **2)** Faça um gráfico de barras contendo os top 5 apps por número de instalação.

Trasformação Strings para int

```python
df_temp = df.copy()
df_temp['Installs'] = df_temp['Installs'].str.replace(',' , '')
df_temp['Installs'] = df_temp['Installs'].str.replace('+' , '')
df_temp['Installs'] = df_temp['Installs'].astype(int)
```

Transforma 'Installs' em int e faz o sort:

```python
apps = df_temp.sort_values(by='Installs', ascending=False,).head(5)
```

Separa a coluna apps e seus installs em diferentes listas:

```python
x_apps = apps['App'].tolist()
y_install = apps['Installs'].tolist()
```

Construindo o grafico:

```python
plt.title('Apps Mais Baixados do Gogle Play')

colors = plt.get_cmap('Accent')(np.linspace(0,0.5, len(x_apps)))

plt.bar(x_apps, y_install, color = colors)
plt.xticks(rotation=45, ha='right')
plt.xlabel('Empresas')
plt.ylabel('Instalações Bilhões')
plt.ylim([1, 1500000000])
plt.show()
```

#### **3)** Faça um Grafico de pizza mostrando as categorias de apps existentes no dataset de acordo com a frequência em que elas aparecem

Agrupando as categorias e contando o número de ocorrencias

```python
df_temp = df.copy()
df_temp = df_temp.groupby('Category').size()
categorias = df_temp.sort_values(ascending=False).to_dict()
```

Separa em duas listas as categorias, e as suas quantidades:

```python
list_x = list(categorias.values())[:len(categorias)]
list_label = list(categorias.keys())[:len(categorias)]
```

Construindo o gráfico:

```python
colors = plt.get_cmap('magma')(np.linspace(0.2, 0.7, len(list_x)))
explode = [0.1 if valor > 500 else 0 for valor in list_x]

wedges, texts, autotexts = plt.pie(list_x, 
                        labels=list_label, 
                        labeldistance=1.1, 
                        radius=1.6, 
                        rotatelabels=1,
                        colors=colors, 
                        explode=explode,
                        autopct='%1.1f%%'
                    )

for text in texts[:-10:-1]:
    text.set_fontsize(6)


plt.title("Distribuição de Categorias", 
          fontsize=16, 
          fontweight='bold', 
          y=1.4)

plt.show()
```

- #### Gêneros dos Apps que tem 'Rating' 5.0: 
---


Tranforma em float as informações da coluna 'Rating', e filtra os apps que tem a avaliação igual a 5.0


```python
df_temp = df.copy()
df_temp['Rating'] = df_temp['Rating'].astype(float)
gen_5 = df_temp.loc[df['Rating'] == 5.0]
gen_5 = gen_5.groupby('Genres').size()
gen_5 = gen_5.sort_values(ascending=False).to_dict()
```

Separa o nome dos gêneros e suas quantidades em duas lista diferentes

```python
list_amt = list(gen_5.values())[:len(gen_5)]
list_gen = list(gen_5.keys())[:len(gen_5)]
```

Define a cor das partes, passa as informações pro gráfico, e formata o titulo e as labels:

```python
colors = plt.get_cmap('YlOrBr')(np.linspace(0.3, 0.9, len(list_amt)))

wedges, texts, autotexts = plt.pie( list_amt, 
                                    labels=list_gen, 
                                    labeldistance=1.1, 
                                    radius=1.7, 
                                    rotatelabels=1,
                                    colors=colors, 
                                    autopct='%1.1f%%'
                                )
for text in texts[:-10:-1]:
    text.set_fontsize(5)


plt.title("Gêneros de Apps com Cinco Estrelas", 
          fontsize=19, 
          fontweight='bold', 
          y=1.4
          )

plt.show()
```
